**ELCAT Bar plot takeaways**

- The majority of languages in the dataset ar either threatened or endangered.
- At risk and awakening languages represent less than ~200 of the 3,000+ languages, creating a heavy class imbalance.

```python
plt.bar(elcat['elcat_endangerment'].value_counts().index, elcat['elcat_endangerment'].value_counts().values, color='#d73e22')
plt.xlabel('Endangerment Level')
plt.xticks(rotation=90)
plt.ylabel('Number of Languages')
plt.title('Most ELCAT languages are threatened or endangered')
plt.show()
```

**ELCAT Boxplot Takeaways**

- Since most languages are spoken in just 1 country, max and average gdp values give us essentially the same distributions. We should probably just keep one. 
- Threatened and endangered languages show very similar patterns across all distributions. The model will likely have a hard time distinguishing between the two categories. Can we find anything that may indicate a seperation between them?
- GDP, language family count and country count show heavy right skews. We should log transform them.
- Dormant, critically endangered and awakening languages show distinct patterns across internet usage, urban population, language family count, and gdp. We will need to explore further if there is overlap between the areas where these features and the areas where the languages are spoken to understand whether or not these features are just a descriptor of the area or the language itself.

```python
elcat_numeric_cols = elcat[['max_gdp_percapita','avg_gdp_percapita','max_internet','avg_internet','max_urban','avg_urban','max_urban_growth','avg_urban_growth','country_count','lang_family_count']]

fig, axes = plt.subplots(nrows=5, ncols=2, figsize= (15, 20))
axes = axes.flatten()

for col, ax in zip(elcat_numeric_cols, axes):
    sns.boxplot(data=elcat, y='elcat_endangerment', x=col, ax= ax)
    plt.tight_layout()
```

**ELCAT bar plot by categorical variables and endangerment levels takeaways**

*Remember that 0 is essentially the "best" score for all of these, showing languages that are spoken by 100,000+ people, transmitted through generations, used in multiple social domains, and the trends show members of the community all speaking this langage with numbers showing either stable or increasing trends. 5 is the opposite of all this.*

- These categories may potentiall help us differentiate at a more granular level. Endangered and threatened are seperated, with endangered falling into a lower category (3) across speaker number, trends, and domains of use, and scoring a 2 for transmission.
- Threatened languages show a higher score than endangered languages across all these categories. 
- Critically endangered languages score 5s more than any other endangerment level.
- Since these traits are compounded together by experts to decide the final language endangerment index level, I'm worried about potentially having multicolinearity in my data. Is there enough variation within each category and endangerment level to find a pattern rather than strict rules that produced the final decision? Notably, the ELCAT lists confidence levels with their final language endangerment level rating, showing that there may not be consistency across languages and these categorical variables may still be useful in predicting language endangerment level, especially between the levels that are very close to each other with other features. 

```python
elcat_categorical_cols = ['speaker_number','speaker_number_trends','transmission','domains_of_use']

fig, axes = plt.subplots(nrows=2, ncols=2, figsize= (15,10))
axes = axes.flatten()

for col, ax in zip(elcat_categorical_cols, axes):
    sns.countplot(data=elcat, y= 'elcat_endangerment', hue=col, ax=ax)

plt.tight_layout()
```
