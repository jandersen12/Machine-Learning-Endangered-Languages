## Data Pipeline

*Research Question: Can publicly available socioeconomic and geographic data approximate language endangerment status without requiring on-the-ground linguistic assessment?*

### Sources

#### Endangered Languages Project (ELP/ELCAT)

Previous versions of this project utilized a CSV file from the Endangered Languages Project (ELP). Since then, ELP has shifted to hosting versions of their data on GitHub through the  ELCAT, a repository where they keep the dataset and each of its independent features as individual CSV files. This new data pipeline will dynamically pull the github repository to build a structured pandas dataframe, rather than working with the CSV from the earlier version of this project.

```python
def extract_language_data(page):
    """
    Extracts data tables from a Wikipedia page and returns a DataFrame.
    """
    url = page.fullurl
    print(f"Extracting data from page: {page.title}: {page.fullurl}")

    response = session.get(url)
    response.raise_for_status()

    # Identify all tables on page
    tables = pd.read_html(response.text)
    print(f"Number of tables found: {len(tables)}")

    # Extract first table (may need to adjust if page structure changes)
    if tables:
        return tables[0]
    else:
        print(f"No tables found on page {page.title}")
        return None
```

### World Bank Cleaning

- Pull the latest value for each country. If no value is available, input NaN
- Collapse World Bank data into one dataframe

```python
def select_wb_value(wb_df, val_col_name):
    """Identify the year columns in the World Bank DataFrames 
    and select the most recent non-null value for each country. """

    year_cols = [col for col in wb_df.columns if re.match(r'YR\d{4}', str(col))]

    # Forward fill values across years and select the most recent non-null value for each country
    wb_df['_latest'] = wb_df[year_cols].ffill(axis=1).iloc[:, -1]

    # Rename resulting columns
    result = wb_df[['economy', 'Country', '_latest']].copy()
    result = result.rename(columns= {'_latest': val_col_name})

    return result
```

## Combine DataFrames

**Strategy:**

Glottolog will serve as the spine for the dataframe merges since it contains 8,000+ languages, both endangered and non-endangered. From there, we will merge the additional dataframes into the glottolog data and fill in the features from the table above language by language.

**Approach:**

- Stage 1
    - Merge ELCAT to Glottolog on glottocode. 
    - Compare ELCAT and Glottolog endangerment labels. Both will be kept for now so that we can filter down to just ELCAT languages for a comparison with our archive models, and then try to build a larger model with the Glottolog languages.
- Stage 2
    - Aggregate world bank data based on country codes
- Stage 3
    - Invert wikipedia "official languages" table into language centric binary 
    - Use name to glottocode lookup to match language names in the wiki dataset to their glottcodes (257 did not match)
    - merge political recognition columns from wiki into master dataframe on new glottocode

### Stage 1: ELCAT + Glottolog

```python
# Merge ELCAT to Glottolog on glottocode
elcat_cols = ['Glottocode', 
                'domains_of_use',
                'speaker_number',
                'speaker_number_trends',
                'transmission',
                'alt_names']

elcat_for_merge = elcat_df[elcat_cols + ['endangerment']].rename(columns={'endangerment': 'elcat_endangerment'})

master_df = glottolog_df.rename(columns={'endangerment': 'glott_endangerment'}
                                ).merge(
                                        elcat_for_merge,
                                       left_on='glottocode',
                                       right_on='Glottocode',
                                       how='left').drop(columns= 'Glottocode')

print(master_df.head())
print(f"Master dataframe shape: {master_df.shape}")
print(f"Languages with Glottolog endangerment: {master_df['glott_endangerment'].notna().sum()}")
print(f"Languages with ELCAT endangerment: {master_df['elcat_endangerment'].notna().sum()}")
print(f"Languages with both: {master_df[['glott_endangerment', 'elcat_endangerment']].notna().all(axis=1).sum()}")
print(f"Languages with neither: {master_df[['glott_endangerment', 'elcat_endangerment']].isna().all(axis=1).sum()}")
```

### Stage 2: Add World Bank Values

```python
def alpha2_to_alpha3(code):
    """Convert alpha-2 country code to alpha-3 using pycountry."""
    try: 
        return pycountry.countries.get(alpha_2=code.strip()).alpha_3
    except:
        return None
```

```python
# Explode country lists to one row per (language, country) pair
lang_country = master_df[['glottocode', 'countries']].explode('countries').copy()
lang_country['economy'] = lang_country['countries'].map(alpha2_to_alpha3)
lang_country.head()
```

### Stage 3: Add Wikipedia political recognition

```python
def split_wiki_languages(text):
    if pd.isna(text):
        return []
    # Strip footnote markers like [1], [a], [b]
    text = re.sub(r'\[[^\]]*\]', '', str(text)).strip()
    
    tokens = text.split()
    result = []
    current = []
    depth = 0
    
    for token in tokens:
        opens = token.count('(')
        closes = token.count(')')
        
        if depth == 0 and opens == 0:
            # Regular word, not inside or starting a parenthetical
            if current:
                result.append(' '.join(current))
            current = [token]
        else:
            # Either opening a parenthetical or already inside one
            current.append(token)
        
        depth += opens - closes
    
    if current:
        result.append(' '.join(current))
    
    return [r.strip() for r in result if r.strip()]
```

```python
wiki_cols = {
    'Official language(s)': 'official',
    'Regional language(s)': 'regional',
    'National language(s)': 'national',
    'Minority language(s)': 'minority',
    'Widely spoken': 'widely_spoken'
    }

lang_flags = {} # {normalized_name: {flag: 1, ...}}

for wiki_col, flag_name in wiki_cols.items():
    if wiki_col not in official_languages.columns:
        continue
    for cell in official_languages[wiki_col].dropna():
        for lang in split_wiki_languages(cell):
            key = lang.strip().lower()
            lang_flags.setdefault(key, {})[flag_name] = 1

wiki_lang_df = pd.DataFrame(lang_flags).T.fillna(0).astype(int)
wiki_lang_df.index.name = 'wiki_name'
wiki_lang_df.reset_index(inplace=True)
```

```python
wiki_lang_df.head(10)
```

```python
# map language names to glottocodes
name_to_glottocode = {
    row['name'].lower().strip(): row['glottocode']
    for _, row in glottolog_df.iterrows()
}

# augment with ELCAT alt_names
for _, row in elcat_df.iterrows():
    gc = row['Glottocode']
    for name in [row['Name']]:
        key = str(name).lower().strip()
        if key not in name_to_glottocode:
            name_to_glottocode[key] = gc
    if pd.notna(row['alt_names']):
        for alias in re.split(r'[;,|]', str(row['alt_names'])):
            key = alias.strip().lower()
            if key and key not in name_to_glottocode:
                name_to_glottocode[key] = gc
```

```python
def wiki_name_to_glottocode(wiki_name, lookup):
    full = wiki_name.lower().strip()
    if full in lookup:
        return lookup[full]
    # Strip parenthetical and retry
    base = re.sub(r'\s*\([^)]*\)', '', full).strip()
    return lookup.get(base)
```

```python
wiki_lang_df['Glottocode'] = wiki_lang_df['wiki_name'].apply(
    lambda name: wiki_name_to_glottocode(name, name_to_glottocode)
)

# Collapse multiple wiki names that resolve to the same glottocode
flag_cols = ['official', 'regional', 'national', 'minority', 'widely_spoken']

wiki_lang_df = (
    wiki_lang_df.groupby('Glottocode', as_index=False)[flag_cols].max()
)
```

### Final cleaning

- drop rows with no target feature
- drop columns not in feature table (see above)
- impute 0 for languages with no political status
- rename columns to match feature table
- impute 0 for isolate languages in lang_family_count
- encode target variable as ordinal integers
- encode macroarea as one-hot
- export dataset

```python
def drop_missing_target(df):
    """Drop rows where the target variable 'glott_endangerment' is missing."""

    df = df.copy()
    
    return df.dropna(subset=['glott_endangerment']).reset_index(drop=True)
```

```python
def drop_unused_columns(df):
    """Drop columns that are not in the defined feature table."""
    
    df = df.copy()
    
    cols_to_drop = ['level', 'iso_code', 'countries', 'family_id', 'latitude', 'longitude', 'alt_names']
    
    return df.drop(columns=[c for c in cols_to_drop if c in df.columns])
```

```python
def impute_binary_flags(df):
    """Replace NaN with 0 for official, national, regional, minority, and widely spoken flags and convert to integers."""

    df = df.copy()

    binary_cols = ['official', 'regional', 'national', 'minority', 'widely_spoken']

    for col in binary_cols:
        df[col] = df[col].fillna(0).astype(int)

    return df
```

```python
def rename_cols(df):
    """Rename columns to match feature table."""

    df = df.copy()
    
    return df.rename(columns={
        'name': 'language_name',
        'family_name': 'lang_family',
        'glott_endangerment': 'endangerment_level',
    })
```

```python
def impute_isolates_count(df):
    """Impute 0 for NaN values in lang_family_count where lang_family is None."""

    df = df.copy()
    
    df['lang_family_count'] = df.apply(
        lambda row: 0 if pd.isna(row['lang_family']) else row['lang_family_count'], 
        axis=1
   )
    return df
```

```python
def encode_target(df):
    """Encode target variable as ordinal integers from 0 to 5."""
    
    ENDANGERMENT_ORDER = {
        'not endangered': 0,
        'shifting': 1,
        'threatened': 2,
        'moribund': 3,
        'nearly extinct': 4,
        'extinct': 5
    }

    df = df.copy()
    df['endangerment_level'] = df['endangerment_level'].map(ENDANGERMENT_ORDER)

    return df
```

```python
def encode_macroarea(df):
    """One-hot encode macroarea column."""

    df = df.copy()

    return pd.get_dummies(df,columns=['macroarea'], prefix='area', dummy_na=False, dtype=int)
```

```python
def export_master(df, path='../data/master_df.csv'):
    """Export cleaned master dataframe to CSV."""

    df.to_csv(path, index=False)
```

```python
clean_df = (master_df
            .pipe(drop_missing_target)
            .pipe(drop_unused_columns)
            .pipe(impute_binary_flags)
            .pipe(rename_cols)
            .pipe(impute_isolates_count)
            .pipe(encode_target)
            .pipe(encode_macroarea)
)

export_master(clean_df)
```

```python
# Export elcat languages into seperate csv
elcat_df_clean = clean_df[clean_df['elcat_endangerment'].notna()].drop(columns=['endangerment_level']).copy()
elcat_df_clean.to_csv('../data/elcat_master_df.csv', index=False)
```
